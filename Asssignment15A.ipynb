```python
# ASSIGNMENT 15A: RIDE-HAILING IN PAKISTAN
import pandas as pd
import matplotlib.pyplot as plt
# Load the dataset
df=pd.read_csv(r"C:\Users\Administrator\Downloads\ride_hailing_dataset.csv")
#Q1: What is the average fare by service type?
print("Q1: ",round(df.groupby('service_type')['fare'].mean(),2))
#Q2: Which city has the most trips?
print("Q2:",df.groupby('city')['trip_id'].count().sort_values(ascending=False).head(1))
#Q3: What is the average distance per trip?
print("Q3:", round(df['distance_km'].mean(), 2))
#Q4: Compare trip ratings by gender.
print("Q4:\n", round(df.groupby('gender')['trip_rating'].mean(), 2))
#Q5: Bar chart of service type vs. average fare.
df.groupby('service_type')['fare'].mean().plot(kind='bar', title='Service Type vs. Average Fare', xlabel='Service Type', ylabel='Average Fare'); plt.show()







```
